## Spacy

Spacy is a popular open-source library for Natural Language Processing (NLP) in Python. It provides various features for text processing, including Named Entity Recognition (NER). NER is the task of identifying and classifying named entities in text into predefined categories such as person names, organizations, locations, dates, etc.

```python
import spacy

# Download the pre-trained NER model (only need to do this once)
# !python -m spacy download en_core_web_sm

# Load the pre-trained NER model
nlp = spacy.load("en_core_web_sm")

# Sample text for NER
text = "Apple is looking at buying U.K. startup for $1 billion."

# Process the text with the NER model
doc = nlp(text)

# Print the named entities found in the text
for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```

## Test with Displacy

Displacy is a visualization tool provided by Spacy that allows you to visualize the results of NER in a more intuitive way. It can be used to display the named entities in a text along with their corresponding labels.

```python
from spacy import displacy

# Visualize the named entities in the text
displacy.render(doc, style="ent", jupyter=True)
```
